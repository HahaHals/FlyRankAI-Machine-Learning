## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/flyrank-bih/flyrank-ml-internship-starter"
REPO_DIR = "flyrank-ml-internship-starter"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)
    subprocess.run([sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt"], check=True)
else:
    # find the repo root from wherever this kernel started
    while not os.path.isdir("data/raw") and os.getcwd() != "/":
        os.chdir("..")

print("Working dir:", os.getcwd())
assert os.path.exists("data/raw/content_refresh_anonymized.csv"), "starter CSV not found — are you at the repo root?"
print("Starter data found. You're ready.")

import pandas as pd, numpy as np
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")
print(df.shape[0], "rows,", df.shape[1], "columns")
df.head(3)

print(f"")

clean_table = df[df["avg_position"] > 0]
corr_RANK = clean_table["avg_position"].corr(clean_table["search_volume"])

print("Correlation ranges from -1 (perfectly opposite) to +1 (perfectly together), with 0 meaning no relationship at all.")
print(f"")

print(f"Correlation between Search Volume (by keywords) and Average Position (on results page): {corr_RANK:.3f}")
print(f"Result is near zero, meaning that the relationship between Search Volume and Average Position is almost negligible and is very weak.")
print(f"Search Volume (keywords) barely predicts the Average Position (ranking) of a page on the results page.")

print(f"")

corr_ENG = df["word_count"].corr(df["engagement_rate"])
print(f"Correlation between Word Count (per article) and Engagement Rate: {corr_ENG:.3f}")
print(f"Result is near zero but on the negative side, meaning that the relationship between Word Count and Engagement Rate is inversely proportional. This denotes that as word count increases, engagement rate very slightly decreases. The  effect on engagement is negligible.")
print(f"Word Count (per article) barely affects the Engagement Rate.")

print(f"")

ctr_by_snippet = df.groupby("content_type")["ctr"].mean()
print(ctr_by_snippet.round(4).to_string())
ctr_by_snippet.plot(kind="bar", title="Mean of Click-Through-Rate by Snippet", ylabel="CTR");
```
